# Exploring statistical values of data

#### Importing Libraries

```python
import pandas as pd
import statsmodels.api as sm
```

#### Reading data

```python
df = pd.read_csv('data.csv')
df.head()
```

```python
df.info()
```

#### Changing Labels into 0, 1

```python
df["Class"] = df["Class"].map({2: 0, 4: 1})
```

#### Extracting Features and Label

```python
X = df.drop(columns=["Class", "Sample code number"])
y = df["Class"]
```

#### Adding Constant column

```python
X = sm.add_constant(X)
```

#### Building model

```python
model = sm.Logit(y, X)
results = model.fit()
```

#### Showing summary for maximal number of columns

```python
print(results.summary())
```

#### Droping column that has highest p value (0.976) and again training model

```python
X = X.drop(columns="Uniformity of Cell Size")
model = sm.Logit(y, X)
results = model.fit()
print(results.summary())
```

#### Again droping column that has highest p value (0.537) and training model

```python
X = X.drop(columns="Single Epithelial Cell Size")
model = sm.Logit(y, X)
results = model.fit()
print(results.summary())
```
