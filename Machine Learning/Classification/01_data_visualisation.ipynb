# Exploring and Visualising Data

#### Importing Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

#### Reading data

```python
df = pd.read_csv('data.csv')
df.head()
```

```python
df.info()
```

#### Ploting Count Plot of Benign and Maligant cases

```python
sns.countplot(x="Class", data=df, hue="Class", legend=False, palette="Set2")
plt.title("Benign/Malignant Cases in the Dataset")
plt.xlabel("Class")
plt.ylabel("Count")
plt.xticks([0, 1], ['Benign', 'Malignant'])
plt.show()
```

#### Ploting mutual correlation heatmap

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df.drop(columns="Sample code number").corr(), annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Matrix of Tumor Characteristics")
plt.show()
```

#### Showing Clump Thickness by Class in a Boxplot

```python
sns.boxplot(x="Class", y="Clump Thickness", data=df, hue="Class", legend=False, palette="Pastel1")
plt.title("Box Plot of Clump Thickness by Class")
plt.xlabel("Class")
plt.ylabel("Clump Thickness")
plt.xticks([0, 1], ['Benign', 'Malignant'])
plt.show()
```

#### Showing Uniformity of Cell Size by Class in a Violinplot

```python
sns.violinplot(x="Class", y="Uniformity of Cell Size", data=df, hue="Class", legend=False, palette="Pastel1")
plt.title("Violin Plot of Uniformity of Cell Size by Class")
plt.xlabel("Class")
plt.ylabel("Uniformity of Cell Size")
plt.xticks([0, 1], ['Benign', 'Malignant'])
plt.show()
```

#### Showing Bare Nuclei by Class in a Violinplot

```python
sns.violinplot(x="Class", y="Bare Nuclei", data=df, hue="Class", legend=False, palette="Pastel1")
plt.title("Violin Plot of Bare Nuclei by Class")
plt.xlabel("Class")
plt.ylabel("Bare Nuclei")
plt.xticks([0, 1], ['Benign', 'Malignant'])
plt.show()
```

#### Showing kdeplot of Uniformity of Cell Size by Class

```python
sns.kdeplot(x="Uniformity of Cell Size", hue="Class", data=df, fill=True, palette="Dark2")
plt.title("KDE Plot of Uniformity of Cell Size by Class")
plt.xlabel("Uniformity of Cell Size")
plt.ylabel("Density")
plt.show()
```
